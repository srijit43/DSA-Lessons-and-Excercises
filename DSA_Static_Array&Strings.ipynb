#Array Operations

```python
a = [1,2,3]
print(a)
```

```python
#insert at the end or append
```

```python
#a.append(4)
print(a)
#complexity O(1)
```

```python
#deleting from the end or poping: O(1)
a.pop()
print(a)
```

```python
#insert not at the end:O(n)

a.insert(2,4)
print(a)
```

```python
#modify an element at a certain position: O(1)

a[3]= 5
print(a)
```

```python
#accessing an element at i: O(1)
print(a[1])
```

```python
#asking if an element is present in the array: O(n)

if 6 in a:
  print("True")
else:
  print("False")
```

#Strings and Operations

```python
#append at the end of the string: O(n)

s = 'hello'

b = s + 'z'

print(b)
```

```python
#check in the string: O(n)
if 'e' in b:
  print("True")
else:
  print("False")
```

```python
#length of a string: =O(1)
len(b)
```
